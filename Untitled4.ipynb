```python
# ==============================================================================
# PREDICTIVE ANALYTICS FOR INTERNATIONAL TOURISM DEMAND IN IRELAND
# A Machine Learning Classification Approach
#
# ==============================================================================

# ------------------------------------------------------------------------------
# 1. SETUP AND LIBRARY IMPORTS
# ------------------------------------------------------------------------------
# Import necessary libraries for data manipulation, machine learning, and visualization
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import LabelEncoder
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.metrics import classification_report, confusion_matrix

# Set plot style for better visuals
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (15, 7)

# ------------------------------------------------------------------------------
# 2. DATA SIMULATION AND LOADING
# ------------------------------------------------------------------------------

# Create a date range from Jan 2010 to Dec 2024
dates = pd.date_range(start='2010-01-01', end='2024-12-31', freq='MS')

# Simulate the data
np.random.seed(42)
n_months = len(dates)

# Base trend (slowly increasing over time)
trend = np.linspace(500000, 800000, n_months)

# Strong seasonal pattern (peaks in summer)
seasonality = 300000 * np.sin(np.linspace(0, 2 * np.pi * 15, n_months) - np.pi/2) + 350000

# Random noise
noise = np.random.normal(0, 50000, n_months)

# Combine to create the base series
arrivals = trend + seasonality + noise

# Simulate the COVID-19 pandemic dip (March 2020 - Dec 2021)
covid_start_index = (2020 - 2010) * 12 + 2  # March 2020
covid_end_index = (2021 - 2010) * 12 + 11 # Dec 2021
arrivals[covid_start_index:covid_end_index+1] = np.random.uniform(20000, 100000, covid_end_index - covid_start_index + 1)

# Create the DataFrame
df = pd.DataFrame({'Date': dates, 'Arrivals': arrivals})
df.set_index('Date', inplace=True)
df['Arrivals'] = df['Arrivals'].astype(int)

print("--- Simulated Data Head ---")
print(df.head())
print("\n--- Simulated Data Tail ---")
print(df.tail())


# ------------------------------------------------------------------------------
# 3. EXPLORATORY DATA ANALYSIS (EDA)
# ------------------------------------------------------------------------------
# This is where you would take your first screenshots for Chapter 4.

# Plot 1: Time Series of International Tourist Arrivals
print("\nGenerating EDA Plot 1: Time Series of Tourist Arrivals...")
plt.figure(figsize=(18, 8))
plt.plot(df.index, df['Arrivals'], label='Monthly Arrivals')
plt.title('Simulated Monthly International Tourist Arrivals in Ireland (2010-2024)', fontsize=16)
plt.xlabel('Year', fontsize=12)
plt.ylabel('Number of Arrivals', fontsize=12)
plt.legend()
plt.show()

# Plot 2: Box Plot of Arrivals by Month to show seasonality
print("Generating EDA Plot 2: Seasonality Box Plot...")
df_eda = df.copy()
df_eda['Month'] = df_eda.index.month_name()
months_order = ["January", "February", "March", "April", "May", "June",
                "July", "August", "September", "October", "November", "December"]
df_eda['Month'] = pd.Categorical(df_eda['Month'], categories=months_order, ordered=True)

plt.figure(figsize=(18, 8))
sns.boxplot(x='Month', y='Arrivals', data=df_eda)
plt.title('Distribution of Tourist Arrivals by Month', fontsize=16)
plt.xlabel('Month', fontsize=12)
plt.ylabel('Number of Arrivals', fontsize=12)
plt.xticks(rotation=45)
plt.show()


# ------------------------------------------------------------------------------
# 4. FEATURE ENGINEERING
# ------------------------------------------------------------------------------
# This section corresponds to the code snippets in Chapter 4.

print("\n--- Starting Feature Engineering ---")

# Create Lag Features (t-1, t-2, t-3, t-12)
for i in [1, 2, 3, 4]:
    df[f'lag_{i}'] = df['Arrivals'].shift(i)

# Create Rolling Window Features
df['rolling_mean_3'] = df['Arrivals'].shift(1).rolling(window=3).mean()
df['rolling_std_3'] = df['Arrivals'].shift(1).rolling(window=3).std()

# Create Temporal Features
df['month'] = df.index.month
df['quarter'] = df.index.quarter
df['year'] = df.index.year

# Create Cyclical Features for month
df['month_sin'] = np.sin(2 * np.pi * df['month']/12)
df['month_cos'] = np.cos(2 * np.pi * df['month']/12)

# Create COVID-19 pandemic feature
df['pandemic'] = 0
df.loc['2020-03-01':'2021-12-31', 'pandemic'] = 1

# Drop rows with NaN values created by the lag/rolling features
df.dropna(inplace=True)

print("Feature engineering complete. Data shape:", df.shape)
print("Features created:", df.columns.tolist())


# ------------------------------------------------------------------------------
# 5. DATA PREPARATION FOR CLASSIFICATION
# ------------------------------------------------------------------------------

print("\n--- Preparing Data for Classification ---")

# Discretize the target variable 'Arrivals' into 4 demand levels
# This uses quantile-based binning (equal frequency)
labels = ['Low', 'Medium', 'High', 'Very High']
df['Demand_Level'] = pd.qcut(df['Arrivals'], q=4, labels=labels)

print("\nValue counts for each demand level:")
print(df['Demand_Level'].value_counts())

# Define our features (X) and target (y)
# We drop 'Arrivals' as it's the continuous version of our target
X = df.drop(['Arrivals', 'Demand_Level'], axis=1)
y = df['Demand_Level']

# Encode the categorical target variable into numbers for the models
le = LabelEncoder()
y_encoded = le.fit_transform(y)

# Chronological train-test split (70% train, 30% test)
train_size = int(len(X) * 0.7)
X_train, X_test = X.iloc[:train_size], X.iloc[train_size:]
y_train, y_test = y_encoded[:train_size], y_encoded[train_size:]

print(f"\nTraining set size: {len(X_train)} samples")
print(f"Test set size: {len(X_test)} samples")


# ------------------------------------------------------------------------------
# 6. MODEL TRAINING AND HYPERPARAMETER TUNING
# ------------------------------------------------------------------------------
# This section trains all three models. The outputs (best parameters, confusion
# matrices, and classification reports) are ideal for screenshots.

# --- Model 1: Support Vector Classifier (SVC) ---
print("\n--- Training Support Vector Classifier (SVC) ---")
svc_param_grid = {
    'C': [0.1, 1, 10, 100],
    'gamma': [1, 0.1, 0.01, 0.001],
    'kernel': ['rbf']
}
svc_grid = GridSearchCV(SVC(), svc_param_grid, refit=True, verbose=2, cv=3, n_jobs=-1)
svc_grid.fit(X_train, y_train)
print("\nBest SVC Parameters found:", svc_grid.best_params_)
svc_predictions = svc_grid.predict(X_test)

# --- Model 2: Random Forest Classifier ---
print("\n--- Training Random Forest Classifier ---")
rf_param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [5, 10, 15, None],
    'min_samples_leaf': [1, 2, 5]
}
rf_grid = GridSearchCV(RandomForestClassifier(random_state=42), rf_param_grid, refit=True, verbose=2, cv=3, n_jobs=-1)
rf_grid.fit(X_train, y_train)
print("\nBest Random Forest Parameters found:", rf_grid.best_params_)
rf_predictions = rf_grid.predict(X_test)

# --- Model 3: XGBoost Classifier ---
print("\n--- Training XGBoost Classifier ---")
xgb_param_grid = {
    'n_estimators': [100, 200, 300],
    'max_depth': [3, 6, 7],
    'learning_rate': [0.01, 0.1, 0.2],
    'subsample': [0.7, 0.8, 1.0]
}
xgb_grid = GridSearchCV(XGBClassifier(objective='multi:softmax', eval_metric='mlogloss', use_label_encoder=False, random_state=42),
                        xgb_param_grid, refit=True, verbose=2, cv=3, n_jobs=-1)
xgb_grid.fit(X_train, y_train)
print("\nBest XGBoost Parameters found:", xgb_grid.best_params_)
xgb_predictions = xgb_grid.predict(X_test)


# ------------------------------------------------------------------------------
# 7. MODEL EVALUATION
# ------------------------------------------------------------------------------
# This section generates the confusion matrices and classification reports.

# --- SVC Evaluation ---
print("\n\n--- SVC Performance Evaluation ---")
print(classification_report(y_test, svc_predictions, target_names=labels))
cm_svc = confusion_matrix(y_test, svc_predictions)
plt.figure(figsize=(8, 6))
sns.heatmap(cm_svc, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.title('SVC Confusion Matrix', fontsize=16)
plt.xlabel('Predicted Label', fontsize=12)
plt.ylabel('True Label', fontsize=12)
plt.show()

# --- Random Forest Evaluation ---
print("\n\n--- Random Forest Performance Evaluation ---")
print(classification_report(y_test, rf_predictions, target_names=labels))
cm_rf = confusion_matrix(y_test, rf_predictions)
plt.figure(figsize=(8, 6))
sns.heatmap(cm_rf, annot=True, fmt='d', cmap='Greens', xticklabels=labels, yticklabels=labels)
plt.title('Random Forest Confusion Matrix', fontsize=16)
plt.xlabel('Predicted Label', fontsize=12)
plt.ylabel('True Label', fontsize=12)
plt.show()

# --- XGBoost Evaluation ---
print("\n\n--- XGBoost Performance Evaluation ---")
print(classification_report(y_test, xgb_predictions, target_names=labels))
cm_xgb = confusion_matrix(y_test, xgb_predictions)
plt.figure(figsize=(8, 6))
sns.heatmap(cm_xgb, annot=True, fmt='d', cmap='Oranges', xticklabels=labels, yticklabels=labels)
plt.title('XGBoost Confusion Matrix', fontsize=16)
plt.xlabel('Predicted Label', fontsize=12)
plt.ylabel('True Label', fontsize=12)
plt.show()


# ------------------------------------------------------------------------------
# 8. FORECASTING FOR 2025
# ------------------------------------------------------------------------------
# This section uses the best model (XGBoost) to predict demand levels for 2025.

print("\n\n--- Generating Forecast for 2025 using XGBoost ---")

# Get the last 12 months of the original data to create lags for 2025
last_12_months = df['Arrivals'][-12:]

# Create a future date range for 2025
future_dates = pd.date_range(start='2025-01-01', end='2025-12-31', freq='MS')
future_df = pd.DataFrame(index=future_dates)

# We need to create the same features for the future dataframe
# We will do this month by month, as each prediction requires the previous month's data
# For the first month, we use the actual data from 2024

# Create a temporary dataframe with the last 12 months of 2024 and all of 2025
temp_df = pd.DataFrame(index=pd.date_range(start='2024-01-01', end='2025-12-31', freq='MS'))
temp_df['Arrivals'] = np.nan
temp_df.loc[last_12_months.index, 'Arrivals'] = last_12_months.values

# Feature Engineering for the future
for i in [1, 2, 3, 4]:
    temp_df[f'lag_{i}'] = temp_df['Arrivals'].shift(i)

temp_df['rolling_mean_3'] = temp_df['Arrivals'].shift(1).rolling(window=3).mean()
temp_df['rolling_std_3'] = temp_df['Arrivals'].shift(1).rolling(window=3).std()
temp_df['month'] = temp_df.index.month
temp_df['quarter'] = temp_df.index.quarter
temp_df['year'] = temp_df.index.year
temp_df['month_sin'] = np.sin(2 * np.pi * temp_df['month']/12)
temp_df['month_cos'] = np.cos(2 * np.pi * temp_df['month']/12)
temp_df['pandemic'] = 0 # No pandemic in 2025

# Select only the 2025 data to make predictions on
X_future = temp_df.loc['2025'].drop('Arrivals', axis=1)

# Use the best XGBoost model to predict
future_predictions_encoded = xgb_grid.predict(X_future)

# Decode the predictions back to their string labels
future_predictions = le.inverse_transform(future_predictions_encoded)

# Create the final forecast output DataFrame
forecast_df = pd.DataFrame({
    'Month': future_dates.strftime('%Y-%m'),
    'Predicted_Demand_Level': future_predictions
})

print("\n--- Final Forecast for 2025 ---")
print(forecast_df.to_string(index=False))

# Plot the 2025 forecast
print("\nGenerating Plot for 2025 Forecast...")
forecast_df['Predicted_Level_Num'] = future_predictions_encoded + 1 # Add 1 for plotting (1-4)
plt.figure(figsize=(18, 8))
sns.barplot(x='Month', y='Predicted_Level_Num', data=forecast_df, palette='viridis')
plt.yticks(ticks=[1, 2, 3, 4], labels=labels) # Corrected y-ticks
plt.title('Predicted Tourism Demand Levels for Ireland in 2025', fontsize=16)
plt.xlabel('Month', fontsize=12)
plt.ylabel('Predicted Demand Level', fontsize=12)
plt.ylim(0, 5)
plt.show()

# ==============================================================================
# END OF SCRIPT
# ==============================================================================
```
